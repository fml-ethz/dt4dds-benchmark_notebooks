```python
%load_ext autoreload
%autoreload 2
import dt4dds_benchmark
```

```python
data = dt4dds_benchmark.analysis.Dataset.combine(*[dt4dds_benchmark.pipelines.HDF5Manager(f'./data/{w}/{s}.hdf5').get_data() for s in (
        'aeon_high',
        'aeon_medium',
        'aeon-motif_high',
        'rs_high',
        'rs_medium',
    ) for w in ('errors',)
])

for c in data.separate_by_parameters(['codec.type', 'codec.name', 'clustering.name', 'clustering.type']):
    c.fit('workflow.overall_rate_motif').plot(title_columns=['codec.type', 'codec.name', 'clustering.name', 'clustering.type']).show()
```

```python
df = data.get_fits_by_group(['codec.type', 'codec.name', 'clustering.name', 'clustering.type'], on='workflow.overall_rate_motif', additional_agg={'code_rate': 'mean'})
df['code_rate'] = df['code_rate'].map('{:.2f}'.format)
df['threshold'] = df['threshold'].fillna(0.5)
df['codec.type'] = df['codec.type'].map({'DNAAeon': 'DNAAeon', 'DNAAeonMotifs': 'motifs_aeon', 'DNARS': 'DNARS',})

fig = dt4dds_benchmark.analysis.plotting.tiered_bar(
    df.sort_values(['codec.type', 'code_rate']),
    "codec.type",
    "code_rate",
    "threshold",
)
fig.update_yaxes(
    title_text='Tolerated error rate',
    tickformat=",.0%",
    range=[0, 0.5],
    dtick=0.25,
)
fig.update_layout(
    width=320,
    height=120,
    margin=dict(l=0, r=2, t=2, b=30),
    showlegend=False,
)


fig = dt4dds_benchmark.analysis.plotting.standardize_plot(fig)
fig.update_xaxes(
    tickfont_size=28/3, 
    tickangle=0,
)
fig.show()
fig.write_image(f'./figures/errors.svg')

# save data
df.to_csv('./figures/errors.csv', index=False)
```

```python
data = dt4dds_benchmark.analysis.Dataset.combine(*[dt4dds_benchmark.pipelines.HDF5Manager(f'./data/{w}/{s}.hdf5').get_data() for s in (
        'aeon_high',
        'aeon_medium',
        'aeon-motif_high',
        'rs_high',
        'rs_medium',
    ) for w in ('dropout',)
])

for c in data.separate_by_parameters(['codec.type', 'codec.name', 'clustering.name', 'clustering.type']):
    c.fit('workflow.dropout_motif').plot(title_columns=['codec.type', 'codec.name', 'clustering.name', 'clustering.type']).show()
```

```python
df = data.get_fits_by_group(['codec.type', 'codec.name', 'clustering.name', 'clustering.type'], on='workflow.dropout_motif', additional_agg={'code_rate': 'mean'})
df['code_rate'] = df['code_rate'].map('{:.2f}'.format)
df['threshold'] = df['threshold'].fillna(1.0)
df['codec.type'] = df['codec.type'].map({'DNAAeon': 'DNAAeon', 'DNAAeonMotifs': 'motifs_aeon', 'DNARS': 'DNARS',})

fig = dt4dds_benchmark.analysis.plotting.tiered_bar(
    df.sort_values(['codec.type', 'code_rate']),
    "codec.type",
    "code_rate",
    "threshold",
)
fig.update_yaxes(
    title_text='Tolerated dropout',
    tickformat=",.0%",
    range=[0, 1.0],
    dtick=0.25,
)
fig.update_layout(
    width=320,
    height=120,
    margin=dict(l=0, r=2, t=2, b=30),
    showlegend=False,
)


fig = dt4dds_benchmark.analysis.plotting.standardize_plot(fig)
fig.update_xaxes(
    tickfont_size=28/3, 
    tickangle=0,
)
fig.show()
fig.write_image(f'./figures/dropout.svg')

# save data
df.to_csv('./figures/dropout.csv', index=False)
```
